```python
import yfinance as yf
```

```python
#Intrinsic Value Calculation
vodacom = yf.Ticker("VOD.JO")
outstandingshares = vodacom.info['sharesOutstanding'] #Fetch the number of outstanding shares
```

```python
#Assumptions
required_rate =0.046386319499999995
perpetual_rate = 0.04
cashflowgrowthrate = 0.03

years = [1, 2, 3, 4]
freecashflow = [ 14387000000, 18944000000, 119918000000, 20185000000 ]#Last 4 years in '000s of ZAR'
```

```python
futurefreecashflow = []
discountfactor = []
discountedfuturefreecashflow = []
```

```python
#Calculate Terminal Value
terminalvalue = freecashflow[-1]* (1+perpetual_rate)/(required_rate-perpetual_rate)
```

```python
print(terminalvalue)
```

```python
#Calculate Discount Factor
for year in years:
    cashflow = freecashflow[-1] * (1+cashflowgrowthrate)**year
    futurefreecashflow.append(cashflow)
    discountfactor.append((1+required_rate)**year)
```

```python
print(discountfactor)
print(futurefreecashflow)
```

```python
for i in range(0, len(years)):
    discountedfuturefreecashflow.append(futurefreecashflow[i]/discountfactor[i])
```

```python
print(discountedfuturefreecashflow)
```

```python
#Calculate the discounted terminal value
discountedterminalvalue = terminalvalue/(1+required_rate)**4
discountedfuturefreecashflow.append(discountedterminalvalue)
```

```python
todaysvalue = sum(discountedfuturefreecashflow)
```

```python
fairvalue = todaysvalue/outstandingshares
```

```python
print("The Fair value of Vodacom is ZAR{}".format(round(fairvalue,2)))
```
